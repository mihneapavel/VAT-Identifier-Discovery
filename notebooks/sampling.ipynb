```python
df_active = df[df["CompanyStatus"] == "Active"].copy()
print(f"Active companies: {len(df_active):,} out of {len(df):,} total")
```

```python
N = len(df_active)
z = 1.96
p = 0.5

def sample_size(N, error):
    n = (N * z**2 * p * (1 - p)) / (error**2 * (N - 1) + z**2 * p * (1 - p))
    return math.ceil(n)

n5 = sample_size(N,0.05)
n7 = sample_size(N,0.07)
n10 = sample_size(N,0.10)

print(f"Population (active companies):{N:,}")
print(f"Sample size for 5%:{n5}")
print(f"Sample size for 7%:{n7}")
print(f"Sample size for 10%:{n10}")
```
